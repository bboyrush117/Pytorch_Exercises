# Matrix Multiplication

Given two matrices:

$$
A =
\begin{bmatrix}
1 & 2 \\
3 & 4
\end{bmatrix}
$$

$$
B =
\begin{bmatrix}
5 & 6 \\
7 & 8
\end{bmatrix}
$$

## Tasks

1. Compute the following:
   - Matrix addition: $A + B$
   - Matrix subtraction: $A - B$
   - Element-wise multiplication: $A \odot B$
   - Matrix multiplication: $AB$

2. Compute the transpose of $A$.

3. Compute the determinant of $A$.

4. Compute the inverse of $A$.

5. Verify the following identity:

$$
AA^{-1} = I
$$

where $I$ is the identity matrix.


```python
A = torch.tensor([[1,2],[3,4]])
B = torch.tensor([[5,6],[7,8]])
Matrix_addition = torch.add(A,B)
Matrix_subtraction = torch.subtract(A,B)
Element_wise_multiplication = A * B
Matrix_Multiplication = torch.matmul(A,B)
```

```python
print(Matrix_addition)
```

```python
print(Matrix_subtraction)
```

```python
print(Element_wise_multiplication)
```

```python
print(Matrix_Multiplication)
```

```python
A_transpose = torch.transpose(A,0,1)
print(A_transpose)
```

Dot Product and Norms

Create:

\[
u = [1, 2, 3]
\]

\[
v = [4, 5, 6]
\]

### Tasks
- Compute:
  - Dot product
  - L1 norm of the dot product
  - L2 norm of the dot product
  - Cosine similarity between the u and v vetors

---


```python
u = torch.tensor([1,2,3], dtype=torch.float32)
v = torch.tensor([4,5,6], dtype = torch.float32)
dot_product = torch.dot(u,v)
L1_norm = torch.norm(dot_product, p=1)
L2_norm = torch.norm(dot_product, p=2 )
import torch.nn.functional as F
cosine_similarity = F.cosine_similarity(u.unsqueeze(0),v.unsqueeze(0))
print("Dot product:", dot_product.item())
print("L1 norm of dot_product:", L1_norm.item())
print("L2 norm of dot_product:", L2_norm.item())
print("Cosine similarity:", cosine_similarity.item())
```

## Eigenvalues and Eigenvectors

Create the matrix:

\[
A =
\begin{bmatrix}
7 & 11 \\
5 & 2
\end{bmatrix}
\]

Tasks:
- Compute the eigenvalues
- Compute the eigenvectors
- Print both results

```python
import torch
A = torch.tensor([[7,11],[5,2]], dtype=torch.float32)
eigenvalues, eigenvectors = torch.linalg.eig(A)
print("\n Eigenvalues: \n", eigenvalues)
print("\n Eigenvectors: \n", eigenvectors)
```
